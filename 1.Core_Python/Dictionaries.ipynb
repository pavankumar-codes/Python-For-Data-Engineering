```python
#Create Dictionaries

hashtable={1:"pavan",2:"Apoorva"} #we use key value pair inside {} to create a dictionary
print(hashtable)
print(type(hashtable))

mydict=dict(name="pavan",age=24,employee="bosch")  #we can use dict() constructor to create a dictionary and here we have to use =
print(mydict)


```

```python
#Access Items
mydict={"name":"pavan","age":24,"employee":"bosch"}

print(mydict["name"]) #we can access using key name
print(mydict.get("age")) #we can access using .get() method

print(mydict.keys())   #we can access all keys using .keys() method
print(mydict.values()) #we can access all values using .values() method
print(mydict.items())  # we can access all key value pairs using .items() method
```

```python
#Add Items
mydict={"name":"pavan","age":24,"employee":"bosch"}

mydict["school"]="srichaitanya tehcno school"  #we can assign a value to a key and add to dict
print(mydict)

mydict.update({"marks":50})  #we can use .update() method to add an item into dict
print(mydict)
```

```python
#Update items

mydict={"name":"pavan","age":24,"employee":"bosch"}

mydict["name"]="Apoorva"  #we can assign a value to a key and update to dict
print(mydict)

mydict.update({"age":50})  #we can use .update() method to update an item into dict
print(mydict)
```

```python
#remove items
mydict={"name":"pavan","college":"KL University","age":24,"employee":"bosch"}

mydict.pop("employee")  #it removes the item for the specified key
print(mydict)


mydict.popitem()
print(mydict)           #it removes the last item in the dict

del mydict["college"]
print(mydict)           #it removes the item in dict without reutrning the value

mydict.clear()
print(mydict)            #it clears the contents of the dictionary

del mydict               #it deletes the dict from the memeory







```

```python
# Looping through dictionary

mydict = {
    "name": "pavan",
    "college": "KL University",
    "age": 24,
    "employee": "bosch"
}

# 1. Loop through keys
for x in mydict:
    print(x, end=" ")

print()

# 2. Loop through values using keys
for y in mydict:
    print(mydict[y], end=" ")

print()

# 3. Loop through keys using keys()
for x in mydict.keys():
    print(x, end=" ")

print()

# 4. Loop through values using values()
for y in mydict.values():
    print(y, end=" ")

print()

# 5. Loop through both keys and values
for x, y in mydict.items():
    print(x, y)
```
