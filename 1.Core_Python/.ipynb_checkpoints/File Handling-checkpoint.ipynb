```python
path=r"C:\20LPA\Python-For-Data-Engineering\demo.txt"  #we include r infront of path because , it specifies its a raw string
with open(path,"r") as file:   #here "r" refers read mode
    data=file.read()           #here .read() reads the content of the file in str format
    print(data)
    print(type(data))
```

```python
path=r"C:\20LPA\Python-For-Data-Engineering\demo.txt"  #we include r infront of path because , it specifies its a raw string
with open(path,"r") as file:   #here "r" refers read mode
    data=file.read(5)           #here .read(5) reads the content of the file in str format till 5 characters
    print(data)
    print(type(data))
```
