```python
#Tuples

mytuple=(1,2,3,4,5,5)
print(mytuple)
print(type(tuple))

mytuple=tuple(("pavan","appu"))  #Tuples can be created using tuple() constructor
print(mytuple)

mytuple=(1,) #we have to create more than 1 item , or "," is required to identif as a tuple
print(mytuple)
print(type(mytuple))
```

```python
#Access Tuples
mytuple=(1,2,3,4,5)
print(mytuple[0]) #since tuples are indexed we can use index
print(mytuple[:]) #we can also use range of idices
print(mytuple[-1]) #we can also use -ve indexing
print(mytuple[-1::-1]) #we can also use -ve range indices
```

```python
#update tuples
mytuple=(1,2,3,4)
#mytuple[0]=0  #Since Tuple is immutable , we cant update,add,remove values in tuple

myList=list(mytuple)
myList[0]=0
mytuple=tuple(myList)
print(mytuple)      #if we really want to change , we have to convert it into List
```

```python
#UnpackingTuple
fruits=("apple","banana","strawberry")  #This is called packing of an tuple
print(fruits)

red,yellow,pink=fruits #This is caled unpacking of a tuple
print(red)
print(yellow)
print(pink)

fruits=("apple","banana","strawberry","dragon","raspberry")  #This is called unpacking of an tuple
red,yellow,*pink=fruits  #if the number of variables is less than number of values,you can add an * to the variable name and variables will be assigned to the variable as a list

print(pink)
print(type(pink))
print(type(red))

```

```python
#Loop Tuples
mytuple=(1,2,3,4)
for num in mytuple:
    print(num)
```
