```python
#Sets
myset={1,2,3,4}    #we can create set using {}
print(myset)
print(type(myset))

myset=set((1,2,3,4))   #we can create a set using set() constructor
print(myset)
print(type(myset))

myset={True,1,False,0}
print(myset)     #Sets doesnt allow duplicates , here 1 and true , 0 and false both are same

```

```python
#Access Set Items
myset={1,2,3,4}
#print(myset[0])   #Sets are unordered , we cant access them using indexing

for num in myset:
    print(num,end=" ")
```

```python
#Add items into set
myset={1,3,5,6}
myset.add(7)   #to insert one element into set 
print(myset)

myset={1,2,6,7}
myset2={5,6,3,8}
myset.update(myset2)  #we can use update() method to insert any iterable into set
print(myset)

myset={1,2,6,7}
mylist=[5,6,3,8]
myset.update(myset2)  #we can use update() method to insert any iterable into set
print(myset)


```

```python
#remove set items
myset={1,2,3,"pavan"}
myset.remove("pavan")  #we can use remove() method to remove the set item
print(myset)

myset.discard("pavan")
print(myset)          #we can use discard() method, it doesnt raise an error even if item is not available in the set

#myset.clear()
print(myset)         #it will clear the set items

myset.pop()
print(myset)     #it removed random element

del(myset)
print(myset)   #it removes entire set
```

```python
#join sets
myset1={1,2,3,4}
myset2={5,6,7,8}
myset3=myset1.union(myset2)  #returns all the elements in both the sets
print(myset1)
print(myset2)
print(myset3)

myset1={1,2,3,4}
myset2={5,6,7,4}
myset3=myset1.intersection(myset2)  #returns only the common elements in both the sets
print(myset1)
print(myset2)
print(myset3)

myset1={1,2,3,4}
myset2={5,6,7,4}
myset3=myset1.difference(myset2)  #returns only the elements in set1 which are not in set2
print(myset1)
print(myset2)
print(myset3)


myset1={1,2,3,4}
myset2={5,6,7,4}
myset3=myset1.symmetric_difference(myset2)  #returns only the elements which are not in both the sets
print(myset1)
print(myset2)
print(myset3)


```
