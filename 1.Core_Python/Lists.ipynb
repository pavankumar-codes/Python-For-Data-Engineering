```python
#Intialize List
myList=[1,2,3,4]
print(myList)

myList1=list((1,2,3,4,5))
print(myList1)
```

```python
#Accesing Elements in Lists
myList=[1,2,3,4]
print(myList[0]) #We can access using Indexing

myList=[1,2,3,4]
print(myList[1:3]) #We can access using range of Indexes

myList=[1,2,3,4]
print(myList[-1::-1]) #We can access using negative inidces and negative range of Indices


```

```python
#Change List Items
myList=[1,2,3,4]

myList[0]=-1 #We can change the List items using indexing
print(myList)

myList[0:1]=[1] #We can change the List items using range of indeces
print(myList)

myList[0:4]=[1]
print(myList) #If you replace a slice with fewer elements than the slice size, the list will shrink.


```

```python
#Add List items
myList=[2,3,4]
myList.append(5)  #we can use append() to insert element at the end
print(myList)


myList.insert(0,1)  #we can use insert() to insert element at the desired position
print(myList)

myList.extend([6,7]) #we can use extend() to insert multiple elements at the end
print(myList)
```

```python
#Remove Items
myList=[2,3,4,5,6,7]
myList.pop()  #pop() allows us delete the element at the end of list
print(myList)

myList.pop(1)
print(myList) #pop("position") allows us to delete the element at any desired position

fruits=['apple','banana','guava']
fruits.remove('apple') #remove() allows us to remove the list items 
print(fruits) 

names=['pavan','appu','maheshbabu']
del names[0] #del(position) allows us to delete the list item and it also allows us to delete the entire list
print(names)

names=['pavan','appu','maheshbabu','darshan']
names.clear()
print(names) #clear() allows us to remove the contents of the list and still an empty list exists
```

```python
#List Comprehensions
myList=[1,2,3,4,5,6]
even_label=["even"for num in myList if num%2==0]
print(even_label)

myList=[1,2,3,4,5,6]
even_label=["even" if num%2==0 else "odd" for num in myList ]
print(even_label)

myList=[1,2,3,4,5,6]
squared=[num*num for num in myList ]
print(squared)

myList=[ [] for i in range(5)]
print(myList)


matrix = [[1,2],[3,4],[5,6]]
output=[matrix[i][j] for i in range(len(matrix)) for j in range(len(matrix[i]))]
print(output)

output=[element for sublist in matrix for element in sublist]
print(output)

words = ["hi", "hello", "hey", "python"]
output=[word for word in words if len(word)>3]
print(output)
```
